```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

subjects = {
    "Math": math,
    "Physics": physics,
    "English": english
}

results = {}

for subject, scores in subjects.items():
    results[subject] = {
        "average": sum(scores) / len(scores),
        "minimum": min(scores),
        "maximum": max(scores)
    }

for subject, result in results.items():
    print(
        subject,
        "→ середнє:", result["average"],
        "мін:", result["minimum"],
        "макс:", result["maximum"]
    )

best_subject = max(
    results,
    key=lambda subject: results[subject]["average"]
)

print("\nПредмет з найвищим середнім результатом:", best_subject)
```
