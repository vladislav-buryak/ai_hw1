```python
from collections import Counter

votes = [
    "apple",
    "banana",
    "apple",
    "orange",
    "banana",
    "apple",
    "apple",
    "apple",
    "orange"
]

counts = Counter(votes)

leader = counts.most_common(1)[0]

print("Кількість голосів:")
for fruit, count in counts.items():
    print(f"{fruit}: {count}")

print("\nЛідер:")
print(f"{leader[0]} — {leader[1]} голосів")
```

```python
top_3 = counts.most_common(3)

print("Топ-3:")
for place, (fruit, count) in enumerate(top_3, start=1):
    print(f"{place}. {fruit} — {count}")
```
